```python
#diagrama de barras

import matplotlib.pyplot as plt
import numpy as np

#comida favorita

comida =["hamburguesa", "perro", "salchipapas", "choriperro", "pizza"]
cantidad=[3,5,8,2,7]

plt.bar (comida,  cantidad,   width=0.5 , color= ["pink", "blue","green", "red","orange" ])
plt.title("Comida Favorita",fontsize=16)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.grid(False)
plt.show()
```
